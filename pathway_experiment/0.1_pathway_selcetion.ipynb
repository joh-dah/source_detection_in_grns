```python
import json

metadata_file = "kegg_data/kegg_nodes_metrics.json"
with open(metadata_file, "r") as f:
    metadata = json.load(f)
```

```python
min_nodes = 80
max_nodes = 90
selected_pathways = []
for pathway_id, data in metadata.items():
    num_nodes = data["number_of_nodes"]
    if min_nodes <= num_nodes <= max_nodes:
        selected_pathways.append((pathway_id, num_nodes))
print("Selected pathways with node counts between", min_nodes, "and", max_nodes)
for pathway_id, num_nodes in selected_pathways:
    # turn Circadian Entrainment (hsa04713) into hsa04713
    id = pathway_id.split("(")[-1].strip(")")
    name = pathway_id.split("(")[0].strip().replace(" ", "_").lower()
    print(f"\"{name}\": \"{id}\",")
```
